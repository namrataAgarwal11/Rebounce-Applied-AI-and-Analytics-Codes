```python
#Personal Expense Tracker

#Maintain a dictionary that stores (Category, Expense_dict) pairs
#Expense_dict = itself a dictionary made of 1)Sub-category 2)Expense
#For a particulary category, acces that category, sum up values in its Expense_dict
#For total expense, count sum across all categories
```

```python
import pandas as pd
import os

class Expense_Tracker :

    def __init__(self):
      if os.path.exists("Expense.csv"):
            self.expense_df = pd.read_csv("Expense.csv")
      else:
            self.expense_df = pd.DataFrame(columns=["Category", "Sub-Category", "Amount", "Description"])
      self.display_menu();


    def display_menu(self) :
      print("1. Add expense")
      print("2. View all expenses")
      print("3. View totals by category")
      print("4. Save and quit")
      option = int(input("choose option"))
      if option == 1:
          self.add_expense()
      elif option == 2 :
          self.view_all()
      elif option == 3 :
          self.view_by_category()
      elif option == 4 :
          self.save_and_quit()
      else :
          print("Option not valid. Choose again")
          self.display_menu()


    def add_expense(self) :
      category = input("Enter category")
      subCategory = input("Enter sub category")
      expense = input("Enter expense")
      desc = input("Enter description")
      new_row = pd.DataFrame(
        [{
            "Category": category,
            "Sub-Category": subCategory,
            "Amount": expense,
            "Description": desc,
        }]
      )
      self.expense_df = pd.concat(
          [self.expense_df, new_row], ignore_index=True
      )
      print("Expense added successfully!")
      self.display_menu()


    def view_all(self):
      print("All Expenses")
      if self.expense_df.empty:
        print("No expenses recorded")
      else:
        print(self.expense_df)
      self.display_menu()

    def view_by_category(self):
        print("Total expenses by Category")
        if self.expense_df.empty:
          print("No expenses recorded yet.")
        else:
          totals = self.expense_df.groupby("Category")["Amount"].sum()
          print(totals)
        self.display_menu()

    def save_and_quit(self):
        self.expense_df.to_csv("Expense.csv", index=False)
        print("Expenses saved to Expense.csv. Goodbye!")
        exit()

tracker = Expense_Tracker()

```
