```python
#Problem 5
password = "olikujgvc54321"
isLong = len(password) >= 8
containsAt = '@' in password
isStrong = isLong & containsAt

print ("Password :", password)
print("Length OK (>=8) :", isLong)
print("Contains '@' :", containsAt)
print("Strong :", isStrong)
```
